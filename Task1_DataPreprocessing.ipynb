```python
# Data Preprocessing Demonstration
# a) Handling Missing Data
# b) Noise Detection and Removal
# c) Identifying Data Redundancy and Elimination

import pandas as pd
import numpy as np

# -----------------------------
# Sample Dataset Creation
# -----------------------------
data = {
    'Age': [22, 25, np.nan, 30, 22],
    'Salary': [25000, 27000, 26000, 1000000, 25000],
    'Department': ['HR', 'IT', 'IT', 'HR', 'HR']
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# -----------------------------
# a) Handling Missing Data
# -----------------------------
print("\nHandling Missing Data:")

# Replace missing values with mean
df['Age'].fillna(df['Age'].mean(), inplace=True)

print(df)

# -----------------------------
# b) Noise Detection & Removal
# -----------------------------
print("\nNoise Detection and Removal:")

# Detect noise using threshold (outlier detection)
# Salary above 100000 is considered noise
df = df[df['Salary'] <= 100000]

print(df)

# -----------------------------
# c) Identifying Data Redundancy
# -----------------------------
print("\nRemoving Redundant Data:")

# Remove duplicate records
df = df.drop_duplicates()

print(df)
```
