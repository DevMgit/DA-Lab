```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Create dataset
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May'],
    'Sales': [200, 250, 300, 280, 350],
    'Profit': [50, 60, 80, 70, 90]
}

df = pd.DataFrame(data)
```

```python
# 1. Bar Chart
plt.figure()
plt.bar(df['Month'], df['Sales'])
plt.title("Monthly Sales (Bar Chart)")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
# 2. Line Chart
plt.figure()
plt.plot(df['Month'], df['Sales'], marker='o')
plt.title("Sales Trend (Line Chart)")
plt.show()

```

```python
# 3. Scatter Plot
plt.figure()
plt.scatter(df['Sales'], df['Profit'])
plt.title("Sales vs Profit (Scatter Plot)")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()

```

```python
# 4. Histogram
plt.figure()
plt.hist(df['Sales'])
plt.title("Sales Distribution (Histogram)")
plt.show()
```
