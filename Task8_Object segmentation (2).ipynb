```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.image import imread
from sklearn.cluster import AgglomerativeClustering
```

```python
# Step 1: Read image
image = imread("image.jpg")
```

```python
# Step 2: Resize (reduce computation)
image = image[:80, :80]
```

```python
# Step 3: Get dimensions
h, w, c = image.shape
```

```python
# Step 4: Create feature vector (x, y, R, G, B)
features = []

for i in range(h):
    for j in range(w):
        pixel = image[i, j]
        features.append([i, j, pixel[0], pixel[1], pixel[2]])

features = np.array(features)
```

```python
# Step 5: Apply hierarchical clustering
model = AgglomerativeClustering(n_clusters=3)
labels = model.fit_predict(features)
```

```python
# Step 6: Reshape to image
segmented = labels.reshape(h, w)
```

```python
# Step 7: Display results
plt.subplot(1,2,1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(segmented, cmap='viridis')
plt.title("Segmented Image (Spatial + Color)")
plt.axis("off")

plt.show()
```
