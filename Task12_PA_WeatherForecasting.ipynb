```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# Create dataset
data = {
    'Day': [1, 2, 3, 4, 5],
    'Temperature': [30, 32, 31, 33, 35]
}

df = pd.DataFrame(data)
```

```python
# Define variables
X = df[['Day']]
y = df['Temperature']
```

```python
# Create model
model = LinearRegression()
```

```python
# Train model
model.fit(X, y)
```

```python
# Predict future temperature (Day 6,7,8)
future_days = np.array([[6], [7], [8]])
predictions = model.predict(future_days)
```

```python
print("Predicted Temperatures:", predictions)
```

```python
# Plot graph
plt.scatter(X, y)
plt.plot(X, model.predict(X))
plt.scatter(future_days, predictions)
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.title("Weather Forecasting")
plt.show()
```
