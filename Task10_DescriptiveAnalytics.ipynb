```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Create dataset
data = {
    'Age': [45, 50, 30, 35, 60],
    'BloodPressure': [130, 140, 120, 125, 150],
    'Cholesterol': [230, 250, 200, 210, 270],
    'Diabetes': [1, 1, 0, 0, 1]
}
```

```python
df = pd.DataFrame(data)
```

```python
# Display dataset
print(df)
```

```python
# Descriptive statistics
print("\nStatistical Summary:")
print(df.describe())
```

```python
# Mean values
print("\nMean Values:")
print(df.mean())
```

```python
# Visualization

# Age distribution
plt.figure()
plt.hist(df['Age'])
plt.title("Age Distribution")
plt.show()
```

```python
# Blood Pressure vs Cholesterol
plt.figure()
plt.scatter(df['BloodPressure'], df['Cholesterol'])
plt.title("BP vs Cholesterol")
plt.xlabel("Blood Pressure")
plt.ylabel("Cholesterol")
plt.show()
```

```python
# Diabetes count
plt.figure()
df['Diabetes'].value_counts().plot(kind='bar')
plt.title("Diabetes Count")
plt.show()
```
