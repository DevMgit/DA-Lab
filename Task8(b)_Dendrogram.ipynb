```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.image import imread
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import linkage, dendrogram
```

```python
# Step 1: Read image (place image.jpg in same folder)
image = imread("image.jpg")
```

```python
# Step 2: Resize image (reduce computation)
image = image[:80, :80]
```

```python
# Step 3: Convert image into pixel data
pixels = image.reshape((-1, 3))
```

```python
# Step 4: Take small sample for dendrogram
sample_pixels = pixels[:100]
```

```python
# Step 5: Generate linkage matrix
Z = linkage(sample_pixels, method='ward')
```

```python
# Step 6: Plot dendrogram
plt.figure(figsize=(10,5))
dendrogram(Z)
plt.title("Dendrogram")
plt.xlabel("Pixels")
plt.ylabel("Distance")
plt.show()
```

```python
# Step 7: Apply hierarchical clustering
model = AgglomerativeClustering(n_clusters=3)
labels = model.fit_predict(pixels)
```

```python
# Step 8: Reshape to image
segmented = labels.reshape(image.shape[0], image.shape[1])
```

```python
# Step 9: Display results
plt.subplot(1,2,1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
```

```python
plt.subplot(1,2,2)
plt.imshow(segmented, cmap='viridis')
plt.title("Segmented Image")
plt.axis("off")

plt.show()
```
